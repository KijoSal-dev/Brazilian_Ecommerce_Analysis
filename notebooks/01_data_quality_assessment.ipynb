### 2.1 Missing Values Analysis

The purpose of this analysis is to identify missing values across the datasets and assess their potential impact on subsequent analysis.

```python
# Combine all datasets into one dictionary

datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Order Payments": order_payments,
    "Order Reviews": order_reviews,
    "Products": products,
    "Sellers": sellers,
    "Category Translation": product_category_name_translation,
    "Geolocation": geolocation
}

# Check missing values in each dataset

for name, df in datasets.items():
    print(f"\n{name}")
    print("-" * 40)
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Calculate the missing-value percentage
for name, df in datasets.items():
    missing = df.isnull().sum()

    missing_percentage = (missing / len(df)) * 100

    missing_summary = pd.DataFrame({
        "Missing Values": missing,
        "Missing Percentage": missing_percentage
    })

    missing_summary = missing_summary[
        missing_summary["Missing Values"] > 0
    ]

    if not missing_summary.empty:
        print(f"\n{name}")
        display(missing_summary.round(2))
```

### 3.1 Investigate Date columns

```python
datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Order Payments": order_payments,
    "Order Reviews": order_reviews,
    "Products": products,
    "Sellers": sellers,
    "Category Translation": product_category_name_translation,
    "Geolocation": geolocation
}

date_columns = {
    "Orders": [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],
    "Order Items": [
        "shipping_limit_date"
    ],
    "Order Reviews": [
        "review_creation_date",
        "review_answer_timestamp"
    ]
}

for name, columns in date_columns.items():
    print(f"\n{name}")
    print("-" * 50)

    df = datasets[name]

    for column in columns:
        converted = pd.to_datetime(df[column], errors="coerce")

        failed = df[column].notna() & converted.isna()

        print(
            f"{column}: "
            f"{failed.sum()} invalid date values"
        )
```

### 3.2 Convert the date columns in our working datasets

```python
cleaned_datasets = {
    name: df.copy()
    for name, df in datasets.items()
}

for name, columns in date_columns.items():
    for column in columns:
        cleaned_datasets[name][column] = pd.to_datetime(
            cleaned_datasets[name][column],
            errors="coerce"
        )

print("Date conversion completed.")
```

### 6. Check primary key uniqueness

```python
primary_keys = {
    "Customers": (customers, "customer_id"),
    "Orders": (orders, "order_id"),
    "Order Items": (order_items, ["order_id", "order_item_id"]),
    "Order Payments": (order_payments, ["order_id", "payment_sequential"]),
    "Order Reviews": (order_reviews, "review_id"),
    "Products": (products, "product_id"),
    "Sellers": (sellers, "seller_id"),
}

for name, (df, key) in primary_keys.items():
    print(f"\n{name}")

    if isinstance(key, list):
        duplicates = df.duplicated(subset=key).sum()
        print("Duplicate composite keys:", duplicates)
    else:
        duplicates = df[key].duplicated().sum()
        print("Duplicate primary keys:", duplicates)

    print("Missing key values:", df[key].isnull().any(axis=1).sum()
          if isinstance(key, list) else df[key].isnull().sum())
```

```python
# Identify orders containing zero-value payment records
zero_payment_orders = zero_payments["order_id"].unique()

# Examine all payment records for those orders
payment_summary = (
    order_payments[
        order_payments["order_id"].isin(zero_payment_orders)
    ]
    .groupby("order_id")
    .agg(
        total_payment_records=("payment_sequential", "count"),
        zero_payment_records=(
            "payment_value",
            lambda x: (x == 0).sum()
        ),
        total_payment_value=("payment_value", "sum"),
        payment_types=(
            "payment_type",
            lambda x: ", ".join(x.unique())
        )
    )
)

print("Distinct orders with zero-payment records:", len(payment_summary))

print("\nOrders with positive total payment:")
print((payment_summary["total_payment_value"] > 0).sum())

print("\nOrders with zero total payment:")
print((payment_summary["total_payment_value"] == 0).sum())

display(payment_summary)
```

```python
# Identify the three orders with zero total recorded payment
zero_total_payment_orders = payment_summary[
    payment_summary["total_payment_value"] == 0
].index

# Retrieve order details
zero_payment_order_details = orders[
    orders["order_id"].isin(zero_total_payment_orders)
][
    [
        "order_id",
        "order_status",
        "order_purchase_timestamp",
        "order_delivered_customer_date"
    ]
]

# Retrieve product and freight details
zero_payment_item_details = (
    order_items[
        order_items["order_id"].isin(zero_total_payment_orders)
    ]
    .groupby("order_id")
    .agg(
        number_of_items=("order_item_id", "count"),
        total_product_price=("price", "sum"),
        total_freight=("freight_value", "sum")
    )
)

# Combine order, payment and item information
zero_payment_investigation = (
    zero_payment_order_details
    .merge(
        zero_payment_item_details,
        on="order_id",
        how="left"
    )
)

display(zero_payment_investigation)
```
